**Write a Python function to read a large CSV file and calculate the average of a numeric column without loading the entire file into memory.**

```python
import pandas as pd
import numpy as np

def process_csv(csv_file, column, chunk_size=1000):
    total_records = None
    sum_records = None

    for chunk in pd.read_csv(csv_file, chunksize=chunk_size):

        total_records += len(pd.to_numeric(chunk[column], errors='coerce').dropna())
        sum_records += np.sum(chunk[column])

    return sum_records/total_records


```

**You have a list of dictionaries representing user data. Show how you'd transform this into a pandas Dataframe and perform basic cleaning.**

```python
import pandas as pd
import numpy as np

def clean_strings(df):

    for column in df.columns:
        df[column] = df[column].str.replace(" ","")
    return df

def clean_email(df):
    invalid_emails = ~df['email'].str.contains('@')

    df.loc[invalid_emails, 'email'] = np.nan

    return df
    

user_data_1 = {'name':'user1','email':'user1@email.com'}
user_data_2 = {'name':'  user2','email': 'user@gmail.com'}
user_data_3 = {'name':'user 3', 'email':'useratemail.com', 'age':'30'}



data = pd.DataFrame(data=[user_data_1,user_data_2,user_data_3])
print(data)
data = clean_strings(data)
data = clean_email(data)

print(data)
```

**Write a python function that reads from an API with proper error handling, retries, and exponential backoff.**

```python
import requests
from tenacity import retry, stop_after_attempt, wait_exponential


@retry(
        stop=stop_after_attempt(3),
        wait = wait_exponential(multiplier=2)
)
def get_data(API_url):
    response = requests.get(API_url)
    response.raise_for_status()

    return response.json()

```
